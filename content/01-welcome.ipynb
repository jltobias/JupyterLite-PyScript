## A tiny geographic dataset

The build bundles `data/nyc-observations.geojson` beside these notebooks. It contains **24 synthetic** observations in NYC with invented hour and value fields. They are not measured taxi trips or health records.

```python
import json
from pathlib import Path

dataset = json.loads(Path('data/nyc-observations.geojson').read_text())
print(dataset['metadata']['provenance'])
print('Observations:', len(dataset['features']))
first = dataset['features'][0]
first
```

GeoJSON coordinate order is **longitude, latitude**. At NYC that is approximately −74, 40.7. The data follows WGS 84 / EPSG:4326. The scene lab changes display coordinates but exports these original coordinates.

```python
lon, lat = first['geometry']['coordinates']
print(f"{first['properties']['name']}: longitude {lon}, latitude {lat}")
assert first['properties']['simulated'] is True
```
